# **Entorno Virtual**

.venv es un entorno virtual de Python.

**Un entorno virtual es una carpeta que contiene:**

- Un intérprete de Python aislado

- Librerías instaladas solo para ese proyecto

- Configuración independiente del Python global del sistema

**Normalmente se llama:**

.venv

venv

env


# **1. El problema original (por qué existen los entornos virtuales)**

**1.1 Python y las librerías**

Python no trae todas las funcionalidades por defecto.

Usamos librerías externas como:

- requests

- flask

- django

- numpy

Estas librerías se instalan con pip.

**1.2 El conflicto sin entornos virtuales**

Supongamos:

- Proyecto A necesita Django 3.2

- Proyecto B necesita Django 4.2

Si instalás librerías globalmente:

            pip install django


**Resultado:**

- Solo puede existir una versión

- Un proyecto rompe al otro

Este problema se conoce como dependency conflict.

# **2. Qué es un entorno virtual (definición formal)**

Un entorno virtual en Python es:

- Un aislamiento del intérprete de Python y sus dependencias, específico para un proyecto.

**En términos simples:**

- Cada proyecto tiene su propio Python

- Sus propias librerías

- Sus propias versiones

- Sin interferir con otros proyectos ni con el sistema

# **3. Qué es realmente un entorno virtual por dentro**

Cuando creás un entorno virtual, Python genera una carpeta (por ejemplo .venv) que contiene:




**Componentes clave**

- python → ejecutable aislado

- pip → gestor de paquetes propio

- site-packages → librerías del proyecto

No usa el Python global salvo como base inicial.

# **4. Cómo se crea un entorno virtual (nivel básico)**

**Paso 1 – Crear el entorno**

        python -m venv .venv


- python → intérprete base

- -m venv → módulo estándar de entornos virtuales

- .venv → nombre del entorno (convención moderna)

**Paso 2 – Activarlo**

Windows

            .venv\Scripts\activate


Linux / macOS

            source .venv/bin/activate


Indicador visual:

            (.venv) usuario@pc

**Paso 3 – Instalar librerías**

            pip install requests


Se instalan solo en ese entorno.

# **5. Qué significa “activar” un entorno virtual (nivel intermedio)**

Activar un entorno virtual significa:

- Cambiar variables de entorno (PATH)

- Priorizar el python y pip del .venv

- NO cambia Python del sistema

Podés verificarlo con:

            which python


o en Windows:

            where python

# **6. Relación con VS Code (lo que suele confundir)**

VS Code:

- Detecta .venv

- Usa automáticamente su Python

- Ejecuta, depura e instala paquetes allí

# **7. Buenas prácticas profesionales**

**7.1 Un entorno por proyecto**

Siempre:

            1 proyecto = 1 entorno virtual


Nunca compartir entornos.

**7.2 No versionar .venv**

Agregar a .gitignore:

            .venv/


El entorno se recrea, no se comparte.

**7.3 Usar requirements.txt**

Guardar dependencias:

            pip freeze > requirements.txt


Restaurar:

            pip install -r requirements.txt

# **9. Problemas comunes (y cómo detectarlos)**

Error típico
            
            ModuleNotFoundError


**Causas:**

- Entorno no activado

- VS Code usando otro Python

- Paquete instalado globalmente

**Solución:**

            pip --version
            python --version


Verificá que apunten al .venv.

# **10. Nivel experto: por qué esto es crítico en producción**

**10.1 Reproducibilidad**

- El mismo entorno en dev, test y prod

- Menos errores “en mi máquina funciona”

**10.2 Seguridad**

- Aislamiento de paquetes vulnerables

- Menor superficie de ataque

**10.3 Escalabilidad**

- CI/CD

- Docker

- Deploys automatizados

Entornos virtuales son la base de todo pipeline serio en Python.

# **11. Concepto final (definición para recordar)**

Un entorno virtual en Python es un aislamiento reproducible del intérprete y dependencias que garantiza estabilidad, compatibilidad y profesionalismo en el desarrollo de software.
