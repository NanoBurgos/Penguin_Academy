**Challenge 2:** Palíndromos con Cola
Crea un programa que determine si una palabra es palíndromo utilizando una lista como cola.

Un palíndromo es una palabra que se lee igual de izquierda a derecha que de derecha a izquierda.
Ejemplos: "oso", "radar".

```python
palabra = input('Ingrese una palabra: ')

lista = list(palabra)

largor = len(lista)

palindromo = True

while palindromo :

  cant = int(largor/2)

  for i in range(cant):
    if lista[i] == lista[cant-i]:
      palindromo = True
      if i == cant:
        break

    else:
      print(f'La palabra {palabra} es una palindromo')
      break

print(f'La palabra {palabra} es una palindromo')
```

```python
palabra = input("Ingrese una palabra: ")
cola = list(palabra)

es_palindromo = True
while len(cola) > 1:
    primero = cola.pop(0)
    ultimo = cola.pop()
    if primero != ultimo:
        es_palindromo = False
        break

if es_palindromo:
    print("palindromo")
else:
    print("no es palindromo")
```

```python
# crear nuestro grafo de amigos BFS

facebook_grafo = {
    "Ana":["Beto","Carlos"],
    "Beto":["Ana","Sancristiano"],
    "Carlos":["Ana","Paolo"],
    "Sancristiano":["Beto"],
    "Paolo":["Carlos"]
}

#imprimimos el grafo completo
print(f"Grafo de amigos de facebook {facebook_grafo}")


# mostrar los amigos de alguien

print(f"Los amigos de Ana son :{facebook_grafo['Ana']}")


def encontrar_amigos_red(grafo,nodo_inicial): # persona inicial = Ana basicamente

    print('Entrando en la funcion')
    cola = [nodo_inicial] # una cola para saber a quien debemos de visitar luego

    lista_visitados = [nodo_inicial] # ana



    print(f"Recorremos la red de amigos de {nodo_inicial}")


    while cola:
        persona_actual = cola.pop(0) # primera persona en ser atendida primero seria ana

        print(f"Estamos visitando a esta persona {persona_actual}")


        # recorrer a la lista de los amigos
        for amigo in grafo[persona_actual]: # en la primera iteracion corresponde a ana
            if amigo not in lista_visitados: # si amigo no fue visitado
                cola.append(amigo)
                lista_visitados.append(amigo)


    return lista_visitados





amigos_encontrados = encontrar_amigos_red(facebook_grafo,"Ana")
print(amigos_encontrados)
```
