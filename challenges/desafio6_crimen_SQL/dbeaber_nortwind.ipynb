# **🔗 NIVEL 2 — JOINs (desde cero)**

🧠 IDEA CLAVE (antes de escribir SQL)

Un JOIN sirve para unir filas de dos tablas relacionadas usando una clave.

En Northwind, las relaciones más importantes son:

- Customers.CustomerID ↔ Orders.CustomerID

- Orders.OrderID ↔ OrderDetails.OrderID

- Products.ProductID ↔ OrderDetails.ProductID

- Products.CategoryID ↔ Categories.CategoryID

**👉 Regla de oro:**

Un JOIN siempre une una PK (Primary Key) con una FK (Foreign Key).

# **🔹 INNER JOIN (el más usado)**

Un INNER JOIN devuelve solo los registros que existen en ambas tablas.

# **🔹 LEFT JOIN (muy importante)**

Un **LEFT JOIN** devuelve:

- Todo lo de la tabla izquierda

- Aunque no tenga coincidencia en la derecha

# **🧠 DIFERENCIA CLAVE**

|JOIN	|Qué devuelve|
|:-:|:-:|
|INNER JOIN	|Solo coincidencias|
|LEFT JOIN	|Todo lo de la izquierda|
|RIGHT JOIN	|SQLite NO lo soporta|
|FULL JOIN	|SQLite NO lo soporta|
