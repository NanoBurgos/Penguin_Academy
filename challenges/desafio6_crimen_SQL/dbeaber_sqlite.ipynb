# **¿Qué es SQLite?**

-SQLite es una base de datos ligera, embebida y autónoma, que funciona dentro de la misma aplicación que la usa.

- No tiene procesos externos, ni puertos, ni servicios en ejecución.

- Solo es un archivo en el disco.

Ejemplo: mi_base.db

Tu programa (Python, JavaScript, Android, etc.) lo abre como si fuera un archivo normal y ejecuta SQL dentro.

# **¿Por qué se usa tanto?**

Porque es:

✅ Súper liviano

El motor completo pesa menos de 1 MB.

✅ Cero configuración

No hay que instalar un servidor ni crear usuarios.

✅ Alta estabilidad

Se usa en millones de aplicaciones.
Está dentro de Chrome, Firefox, WhatsApp, Android, iOS, Windows, Tesla, aviones…
Es famoso por ser muy robusto.

✅ Portátil

La base de datos es solo un archivo → podés copiarlo, moverlo, subirlo a Git, hacer backup arrastrándolo a una carpeta.

✅ Compatible con SQL estándar

Las consultas son casi iguales a las de otros motores SQL.

# **¿Para qué usar SQLite?**

Ideal para:

- Aplicaciones móviles (Android e iOS)

- Proyectos pequeños o medianos

- Programas de escritorio

- Web scraping (guardar datos scrapeados)

- Aprender SQL

- Apps que no necesitan múltiples usuarios escribiendo al mismo tiempo

- Prototipos rápidos

# **¿Cuándo NO conviene usar SQLite?**

No es ideal si:

- Muchas personas o procesos van a escribir al mismo tiempo.

- La base es muy grande (decenas o cientos de GB).

- Necesitás roles avanzados, replicación, clustering, alta concurrencia.

Para eso conviene Postgres o MySQL.

# **¿Cómo guarda los datos?**

En un único archivo binario.

Ejemplo:

            mi_base.db


Ese archivo contiene tablas, índices, triggers, datos, todo.

# **¿Qué limita SQLite?**

❗ Concurrencia

- Muchas lecturas simultáneas no son problema.

- Muchas escrituras simultáneas sí pueden generar bloqueos (database is locked).

❗ Tipado flexible

- SQLite tiene “dynamic typing”:

El tipo declarado en una columna es más una sugerencia que una obligación.

Ejemplo:
Podés insertar texto en una columna INTEGER (aunque no es recomendable).

# **🔹 ¿Cómo se usa con Python? (ejemplo práctico)**

```python
import sqlite3

# Crear o conectar a un archivo de base de datos
conn = sqlite3.connect("mi_base.db")
cursor = conn.cursor()

# Crear tabla
cursor.execute("""
CREATE TABLE IF NOT EXISTS personas (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre TEXT,
    edad INTEGER
)
""")

# Insertar datos
cursor.execute("INSERT INTO personas (nombre, edad) VALUES (?, ?)", ("Juan", 25))

conn.commit()

conn.close()
```

# **¿Cuando ya no voy a usar dbeaber, debo cerrar la conexion?**

**En DBeaver no es obligatorio cerrar la conexión** cada vez que dejás de usarla, pero sí es una buena práctica, especialmente en estas situaciones:

# **✅ Cuándo conviene cerrar la conexión**

- Cuando vas a cerrar DBeaver completamente.

- Cuando estás conectado a una base de datos remota (por ejemplo, un servidor externo).

- Evita mantener sesiones abiertas innecesariamente.

- Cuando usás bases de datos que tienen límite de conexiones (PostgreSQL, MySQL, etc.).

- Cuando vas a cambiar credenciales, mover el proyecto o exportarlo.

# **❌ Cuándo NO hace falta cerrar la conexión**

- Si solo dejás la ventana abierta mientras trabajás.

- Si vas a ejecutar consultas de manera frecuente.

- Si la base es local (SQLite, un archivo), ya que no mantiene una sesión viva como las bases cliente-servidor.

# **¿Qué pasa si no cierro la conexión?**

Generalmente nada grave:

- El servidor podría dejar la conexión inactiva y cerrarla solo.

- Algunas bases tienen un timeout automático (5 min, 1 h, depende de la configuración).

- Podrías ver un mensaje de conexión expirada cuando querés ejecutar una consulta, y DBeaver reconecta solo.

# **⭐ Recomendación práctica**

Si trabajás en una base remota o con límite de sesiones:

👉 Menú derecho sobre la conexión → Disconnect cuando termines.

Si es una base local o de un curso, no pasa nada si la dejás abierta.
