# **DOM**

**DOM (Document Object Model)** : Interfaz independiente del lenguaje que trata un documento XML o HTML como una estructura de arbol



# **Etiquetas**

- **Estructura Raíz:** < html >, < head >, < body >.

- **Encabezados/Titulos:** < h1 > , < h2 >,..., < h6 >   **(siendo < h1 > el más importante)**

- **Contenido:** < p > (párrafo), < a > (enlace), < img > (imagen), < div > (contenedor), <span>.

< href > referencia a la direccion de la pagina que queremos vincular

- **Listas:** < ul > (desordenada), < ol > (ordenada), < li > (elemento de lista).

- **Formularios:** < form >, < input >, < button >, < label >

- **Semánticas (HTML5):** < header >, < nav >, < article >, < section >, < footer >. 

Las etiquetas pueden tener atributos, como src para imágenes o href para enlaces, que añaden información extra.
