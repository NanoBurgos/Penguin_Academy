## **FUNCIONES DE VENTANA**

Las funciones de ventana en SQL son **herramientas analíticas avanzadas que realizan cálculos sobre un conjunto de filas relacionadas con la fila actual (una "ventana"), sin agruparlas en una sola salida.** 

A diferencia de GROUP BY, mantienen el detalle individual mientras calculan totales acumulados, clasificaciones o promedios móviles. 

Aspectos Clave y Funcionalidad:
- Cláusula **OVER()**: Es el núcleo de estas funciones, definiendo el marco de datos sobre el que operan.
- Particionamiento y Orden: **PARTITION BY** divide los datos en grupos, mientras que **ORDER BY** organiza las filas dentro de cada grupo.
- Diferencia con Agregación: Las funciones de agregación (como **SUM**, **AVG**) devuelven un solo valor por grupo, mientras que las de ventana devuelven un valor para cada fila analizada. 

Tipos Comunes de Funciones de Ventana:
- Ranking: **ROW_NUMBER(), RANK(), DENSE_RANK()**.
- Valor: **LAG(), LEAD()** (para comparar con filas anteriores/siguientes), **FIRST_VALUE(), LAST_VALUE()**.
- Agregación: **SUM(), AVG(), COUNT()** usadas con **OVER()**. 

Casos de Uso Principales:

- Totales Acumulados: Calcular la suma de ventas hasta la fecha actual.
- Ranking: Clasificar empleados por salario dentro de cada departamento.
- Análisis Comparativo: Comparar las ventas de este mes con las del mes anterior.
