requests → para descargar el HTML de una página web

# **Cual seria la parte ILEGAL?** (lo que nunca deberías hacer).

❌ 1. Scraping datos privados o no accesibles públicamente

Ejemplos ilegales:

Scraping un sitio detrás de login (Facebook, Instagram, bancos)

Scraping datos personales sin permiso (nombres, emails, teléfonos)

Scraping datos vendidos por suscripción de pago

👉 Si hace falta iniciar sesión, no lo scrapees.

❌ 2. Ignorar el archivo robots.txt

Este archivo indica qué partes del sitio permiten scraping.

Si el sitio dice:

      User-agent: *
      Disallow: /


Significa: prohibido scrapear todo.

Ignorarlo puede violar Términos de Servicio.

❌ 3. Hacer demasiadas requests (ataque DDoS accidental)

Scraping mal hecho puede tumbar un servidor pequeño.

Ejemplo ilegal o dañino:

      for i in range(1_000_000):
          requests.get(url)   # sin pausa


👉 Mandar demasiadas peticiones demasiado rápido puede ser visto como un ataque.

❌ 4. Saltarse medidas de seguridad

Prohibido:

Evitar CAPTCHAs con bots

Romper un firewall

Usar proxys para eludir bloqueos

Hacer ingeniería inversa para acceder a APIs privadas

👉 Todo eso puede ser judicialmente perseguible.

❌ 5. Usar scraping para fines maliciosos

Por ejemplo:

Robar precios para copiar un sitio competidor

Crear listas de emails para spam

Extraer información personal para phishing

Obtener ventajas comerciales sin autorización

❌ 6. Copiar contenido protegido por derechos de autor

Si scrapear un sitio implica:

Bajar masivamente textos de libros

Descargar cursos

Descargar películas o PDFs pagos

Eso es ilegal incluso si la técnica es válida.

❌ 7. Violación de Términos y Condiciones

Aunque técnicamente posible, es ilegal si el sitio lo prohíbe expresamente.

Ejemplo (imaginario):

“El uso de bots o software de scraping está prohibido.”

Si igual haces scraping → incumplimiento contractual.

# **Cada pais tiene leyes diferentes en cuanto al scraping?**

🌍 1. Europa – GDPR (Reglamento General de Protección de Datos)

La Unión Europea tiene las leyes más estrictas del mundo sobre datos personales.

Scraping es ilegal si extraés:

datos identificables (email, nombre, dirección)

sin consentimiento del usuario

o sin base legal válida

Incluso si los datos eran visibles en la web.

➡️ En Europa, scrapear datos personales es casi siempre ilegal.

Pero scraping de datos no personales (libros, precios, productos) suele ser legal.

🇺🇸 2. Estados Unidos – más permisivo

En EE.UU., la ley es menos estricta.

Hay un caso famoso: hiQ vs LinkedIn (2022)
La corte decidió que scrapear información pública NO es ilegal, incluso si un sitio lo prohíbe en su Términos de Servicio.

Eso hizo que EE.UU. tenga una postura más “scraping-friendly”.

➡️ Pero sigue siendo ilegal:

entrar a áreas privadas

evitar sistemas de seguridad

scrapear con intención maliciosa

🇦🇷 🇵🇾 🇨🇱 🇺🇾 – Latinoamérica

En América Latina, la mayoría de países tienen leyes inspiradas en GDPR, pero menos estrictas, por ejemplo:

Argentina – Ley 25.326

Prohíbe recolectar datos personales sin consentimiento.
Pero scraping de precios, productos, libros, etc., es permitido.

Paraguay – Ley 6534/20

Protege datos personales sensibles.
Scraping de datos no personales = permitido.
Datos personales = solo con consentimiento.

Chile – Ley 19.628

Similar: protege datos sensibles.
No prohíbe scraping de datos no personales.

En general en LATAM:

Scraping de datos personales sin permiso → ❌ ilegal

Scraping de datos generales públicos → ✔️ legal

🇯🇵 Japón – APPI

Muy parecido a GDPR.
Datos personales → protegidos.
Scraping no personal → permitido.

🇦🇺 Australia

Protege datos personales, pero no castiga scraping de datos públicos mientras no rompas seguridad.

# 🧠 **Entonces… ¿qué importa realmente?**

Para saber si el scraping es legal tenés que mirar tres cosas:

✔️ 1. ¿Los datos son personales?

Si sí → probablemente ilegal sin consentimiento.

✔️ 2. ¿Los datos son públicos sin login?

Si sí → en muchos países el scraping es legal.

✔️ 3. ¿Rompiste seguridad, saltaste login o cargaste el servidor?

Si sí → ilegal en casi todo el mundo.
