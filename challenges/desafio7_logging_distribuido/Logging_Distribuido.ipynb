**Logging Distribuido en Microservicios | Primer Pilar de la Observabilidad (Explicado Simple)**

Que es Logging?

- Una forma sencilla de proporcionar informacion sobre el estado de la aplicacion

- Una linea de Log puede ser:

1. Evento

2. Accion

3. Excepcion

4. Error

Las Lineas de Log deben de ser faciles de leer por humanos pero tambien faciles de interpretar por maquinas para que puedan ser analizadas, agrupadas y procesadas eficientemente


image.png

## **Loggin en Arquitectura de Microservicios**

Podemos tener miles de instancias de diferentes microservicios produciendo colectivamente millones de lineas de **log** por dia; y cuando tengamos un problema no sera practico hacer ssh y revisa esos logs uno por uno.

Por eso la 1era Practica Recomendada para poder darle sentido a esa enorma cantidad de datos, es recopilarlos en un **Sistema de Logging Centralizado**

Este sistema debe de analizar e indexar esos logs para que puedan ser buscados facilmente por patrones de texto agrupados o filtrados por atributos como: HOST, MICROSERVICIO, PERIODO DE TIEMPO o REGION

La 2da Practica Recomendada es **Asignar un nivel de log o severidad a cada linea.**

Dependiendo del Framework o sistema que usemos, existen diferentes niveles pero las mas conocidas son:

- TRACE
- DEBUG
- INFO
- WARN
- ERROR
- FATAL

image.png


Esto agrega una dimension adicional a los logs, que permite filtrarlos y reducir ruido o alertas inecesarias

**Por ej:** las excepciones y los errorres que afectan a los usuarios deben registrarse como ERROR Y FATAL; asi si un ingeniero de guardia recibe una alerta en la madrugada indicando que algo esta fallando, puede buscar en los logs filtrando solamente por ERROR Y FATAL, descubrir que paso exactamente y tomar las acciones correctas,

Tambien se pueden tener herramientas automaticas que escaneen periodicamente y agrupen eventos por severidad y que generen alertas o creen tickets asignados automaticamente

image.png

La siguiente Practica Recomendada es **usar un ID Unico** comunmente llamado **Correlation ID** para cada request o transaccion del usuario y agregar este ID a cada linea de Log correspondiente a ese proceso, dado a que cada instancia de microservicio probablemente esta procesando multiples solicitudes simultaneamente,

- Esto ayuda a buscar y filtrar los eventos relacionados a un request especifico
image.png


- Tambien nos permite ver la secuencia de eventos de una solicitud a traves de varios microservicios
image.png

Otra buena practica es **Agregar la mayor informacion contextual posible a cada linea de log**

**Por ej:** Si registramos un error o excepcion queremos incluir el stack trace para entender como llegamos a esa linea de codigo, asi como los parametros que llevaron a el error

image.png

- Si tenemos una consulta de base de Datos que tardo demasiado tiempo, registrar la consulta exacta y el contenido de la request que la disparo, puede ayudarnos a entender como ocurrio el problema y como mitigarlo

## **Datos comunes que debemos incluir en cualquier linea LOG**
- Nombre del Servicio
- Hostname / Direccion IP (nombre del host donde ocurrio el evento)
- User ID (nombre de usuario o cualquier otro identificador util)
- Timestamp

# **Informacion Contextual - Consideraciones**

1. Registrar solo los datos necesarios (solo informacion critica para depuracion)(En sistemas a gran escala el almacenamiento y proceso puede ser muy costoso)

2. No Registrar: Datos sensibles, Informacion de identificacion personal (Nombres reales, contraseñas, Numero de seguro social, correo, numero de tarjeta de credito, etc) (Representan un riesgo legal enorme en caso de una brecha de seguridad)
