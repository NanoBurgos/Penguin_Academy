## **API REST**

**Una API REST (o RESTful API)** es una interfaz de programación que permite que dos sistemas se comuniquen entre sí utilizando el protocolo **HTTP**, siguiendo los principios de la arquitectura **REST (Representational State Transfer)**, definida por Roy Fielding en su tesis doctoral.

## 1. **¿Qué es una API?**

API (Application Programming Interface) es un contrato que define:

- Qué operaciones se pueden realizar

- Cómo deben enviarse los datos

- Qué formato tendrá la respuesta

Ejemplo:

Un frontend en React consume una API para obtener libros desde un backend en Python.

## 2. **¿Qué significa REST?**

REST es un estilo arquitectónico, no un protocolo ni una tecnología.

**Se basa en:**

**a) Recursos**

Todo es un recurso identificado por una URL.

Ejemplo:

            /usuarios
           
            /libros
           
            /pedidos/15

**b) Uso correcto de métodos HTTP**

|Método|	Acción|
|:-:|:-:|
|GET|	Obtener datos|
|POST|	Crear|
|PUT|	Actualizar completo|
|PATCH|	Actualización parcial|
|DELETE|	Eliminar|

Ejemplo práctico:

            GET /libros
           
            POST /libros
           
            GET /libros/10
           
            DELETE /libros/10

**c) Stateless (Sin estado)**

Cada request debe contener toda la información necesaria.
El servidor no guarda estado de la sesión entre peticiones.

Ejemplo:

Si usás JWT, el token va en cada request.

**d) Representaciones**

Los datos se intercambian generalmente en JSON.

Ejemplo respuesta:

            {
          
            "id": 1,
          
            "titulo": "El Principito",
          
            "precio": 50000
          
            }

## 3. **¿Qué hace que una API sea RESTful?**

Para ser RESTful debe:

- Usar correctamente HTTP

- Ser stateless

- Exponer recursos, no acciones

- Tener URLs claras

- Usar códigos de estado HTTP apropiados

**Ejemplo correcto:**

            POST /usuarios

**Ejemplo incorrecto (no REST):**

            /crearUsuario

## 4. **Ejemplo real**

Cuando usás la API de Google Maps o la API de OpenAI, estás consumiendo APIs REST que responden con JSON vía HTTP.

## 5. **En tu contexto (backend con base de datos)**

**Una API REST sería:**

Backend:

- Conecta a la base de datos

- Expone endpoints

- Devuelve JSON

Frontend:

- Hace fetch o axios

- Consume datos

## **Resumen técnico corto**

Una API REST es:

- Una interfaz basada en HTTP que expone recursos mediante URLs, usa métodos HTTP estándar y es stateless.
