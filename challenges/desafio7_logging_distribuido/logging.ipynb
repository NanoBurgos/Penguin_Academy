# **ETAPA 1 — Fundamentos conceptuales (Arquitectura)**

Antes de escribir una sola línea de código

**1.1 ¿Qué es un sistema de logging distribuido?**

Tenés que entender:

- Qué es un log

- Para qué sirve en sistemas reales

- Diferencia entre:

Logs locales

Logs centralizados

Logging distribuido

**Conceptos clave a estudiar:**

Event logging

Centralized logging

Observability (logs vs metrics vs traces)

Por qué los logs no deben perderse


**Objetivo:**

Poder explicar con tus palabras por qué este sistema existe.


**1.2 Arquitectura del desafío**

Aprender a pensar el sistema como componentes, no como código.
Componentes:

Servicios simulados (clientes)

Servidor central

Base de datos

Autenticación

API HTTP


**Diagramas mentales:**

Flujo de un log desde que se genera hasta que se consulta

Qué pasa si algo falla (token inválido, DB caída, payload mal formado)

**Objetivo:**

Poder dibujar el sistema en una hoja sin código.



# **ETAPA 2 — HTTP & APIs REST (base obligatoria)**

Este desafío es 80% API bien pensada

**2.1 HTTP en profundidad**

No solo “GET y POST”.

**Tenés que dominar:**

- Métodos HTTP: GET, POST

- Headers (Authorization, Content-Type)

- Status codes:

200, 201

400, 401, 403

500


**Especial foco en:**

- Qué es un request

- Qué es un response

- Idempotencia

- Payload JSON


**Objetivo:**

Saber exactamente qué entra y qué sale del servidor.


**2.2 Diseño de endpoints**

Diseño limpio y coherente:

POST /logs

GET /logs?timestamp_start=...&timestamp_end=...

Conceptos a estudiar:

RESTful APIs

Query parameters

Validación de datos

Respuestas consistentes

**Objetivo:**

Diseñar endpoints sin escribir código.



# **ETAPA 3 — Autenticación simple pero correcta**

Seguridad básica, sin magia negra

**3.1 Tokens manuales**

Aprender:

- Qué es un token

- Por qué se usa

- Diferencia entre:

- Token simple

JWT (aunque acá no lo uses)


**Conceptos clave:**

Authorization header

Validación del token

Respuestas ante token inválido


**Objetivo:**

Entender que la autenticación es una barrera, no un adorno.


**3.2 Manejo de errores**

Errores bien hechos:

Error por token inválido

Error por payload mal formado

Error interno

**Aprender:**

- Mensajes claros

- Status codes correctos

- No filtrar info sensible


**Objetivo:**

Que los errores ayuden a debuggear, no a confundir.




# **ETAPA 4 — Persistencia y base de datos**

Acá el log deja de ser humo

**4.1 Modelado de datos**

Antes de usar cualquier DB:

Pensar el esquema del log:

- id

- timestamp (evento)

- received_at (cuando llega)

- service

- severity

- message


**Conceptos:**

Diferencia entre timestamp del evento y del servidor

Índices por fecha

Normalización simple

**Objetivo:**

Saber exactamente qué se guarda y por qué.

**4.2 Base de datos**

Podés usar:

SQLite (ideal para empezar)

PostgreSQL

MySQL


**Aprender:**

INSERT

SELECT con filtros

WHERE por rango de fechas

ORDER BY

**Objetivo:**

Consultar logs como si fuera un sistema real.




# **ETAPA 5 — Servicios simulados (clientes)**

Acá entra la diversión controlada

**5.1 Generación de logs falsos**

Aprender a:

- Generar timestamps reales

- Usar niveles de severidad

- Crear mensajes variados

**Conceptos:**

Random controlado

Simulación realista

Batch de logs (listas)


**Objetivo:**

Que los logs parezcan reales, no de tutorial.


**5.2 Envío HTTP desde scripts**

Aprender:

Requests HTTP desde código

Headers personalizados

Envío de JSON

Manejo de errores de red

**Objetivo:**

Que múltiples servicios puedan convivir sin romper nada.




# **ETAPA 6 — Performance, orden y criterio**

Nivel “ingeniero”, no solo “programador”

**6.1 Manejo de múltiples logs**

Aprender:

Enviar varios logs en una sola request

Validar listas

Procesar sin bloquear

**Conceptos:**

Batch processing

Validaciones parciales

Respuesta agregada

**Objetivo:**

Que 1000 logs no hagan explotar tu servidor.


**6.2 Código mantenible**

Aprender:

Separación de responsabilidades

Comentarios útiles

Nombres claros

Estructura de carpetas

**Objetivo:**

Que tu “yo del futuro” no te odie.

Orden recomendado de estudio (resumen)

Arquitectura y concepto de logging

HTTP y APIs REST

Autenticación por token

Modelado y base de datos

Servicios simulados

Optimización y limpieza de código
