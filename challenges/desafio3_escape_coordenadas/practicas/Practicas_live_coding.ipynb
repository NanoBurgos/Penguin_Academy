#Practicas Tkinter

```python
import tkinter as tk

#creamos una ventana principal
ventana = tk.Tk() //creamos la ventana principal
ventana.title("Mi primera ventana en Tkinter") //le cambiamos el titulo a la ventana

#ponemos una descripcion como etiqueta
descripcion = tk.Label(ventana, text="Aqui la descripcion") //ventana indica en que ventana se va a mostrar, text="...." indica el contenido de la etiqueta
descripcion.pack(pady=10) //muestra la etiqueta en la ventana usando el gestor de geometria pack()
//pady=10 añade un margen vertical 10 pixeles arriba y abajo

#Funcion que se ejecuta al presionar un boton
def accion_boton():
  descripcion.config(text="Haz presionado el boton")

#Creamos un boton
boton= tk.Button(ventana, text="Presioname", command = accion_boton) //command ->accion que se ejecuta cuando se hace click en el boton
boton.pack(pady=10)

#Mantiene la ventana abierta
ventana.mainloop()
//Inicia el bucle principal de la interfaz gráfica.
//Este bucle se queda “escuchando” eventos (clics, teclas, redibujos).
//Sin esta línea, la ventana se abriría y se cerraría instantáneamente.
```

#Crear tablero con 0's, marcar entrada y salida, imprimir tablero con numeros y tablero lindo

```python
def crear_tablero(filas,columnas):
    for fila in range(filas):
        fila = []
        for _ in range(columnas):
            fila.append(0)

        tablero.append(fila)

    x= 0,0
    y=filas-1, columnas-1

    agregar_inicio_fin(x,y)

def imprimir_tablero():

    for fila in tablero:
        print(fila)

def agregar_inicio_fin(inicio, fin):
    tablero[inicio[0]][inicio[1]]= 1
    tablero[fin[0]][fin[1]]= 2



def imprimir_tablero_lindo(x, y):
    for fila in range(x):
        for columna in range(y):
            if tablero[fila][columna] == 0:
                tablero[fila][columna] = " "
            if tablero[fila][columna] == 1:
                tablero[fila][columna] = "E"
            if tablero[fila][columna] == 2:
                tablero[fila][columna] = "S"


filas = 4
columnas = 6

tablero = []

crear_tablero(filas, columnas)

imprimir_tablero()

imprimir_tablero_lindo(filas, columnas)

imprimir_tablero()
```
