# Para poder resolver este desafio tenemos varios tipos de algoritmos de busqueda

BFS (Breadth-First Search)

Dijkstra

A* (A-star)

# ¿Cual es la diferencia entre cada uno, PROS y CONTRAS?

    BFS (Breadth-First Search): Encontrar el camino más corto en un grid cuando todos los movimientos tienen el mismo costo.

    Dijkstra: Extiende BFS para cuando los caminos tienen diferentes costos (ej: tierra, agua).

    A* (A-star): Algoritmo más eficiente para buscar rutas usando heurísticas.

# BFS (Breadth-First Search)

📌 Idea: explora el mapa nivel por nivel (como una ola que se expande desde el inicio).

Usa una cola (queue).

Visita todos los nodos a distancia 1, luego a distancia 2, etc.

Cuando llega al destino, ese camino es el más corto en cantidad de pasos.

✅ Cuándo usarlo:

Cuando todos los movimientos tienen el mismo costo (ej: cada paso vale 1).

🛠️ Ejemplo en un grid:
Moverse arriba/abajo/izquierda/derecha en un mapa sin pesos.

# Dijkstra

📌 Idea: parecido a BFS, pero considera costos diferentes en los caminos.

Usa una cola de prioridad (priority queue o heap).

Siempre expande el nodo con menor costo acumulado hasta ese punto.

Encuentra el camino más barato, no solo el más corto.

✅ Cuándo usarlo:

Cuando los movimientos tienen pesos distintos (ej: cruzar agua cuesta 3, tierra cuesta 1).

Ejemplo: mapas con terrenos de distinta dificultad.

# A* (A-star)

📌 Idea: mejora Dijkstra usando una heurística que guía la búsqueda hacia el destino.

Usa una función de evaluación:

𝑓(𝑛)=𝑔(𝑛)+ℎ(𝑛)

f(n)=g(n)+h(n)

g(n) = costo real desde el inicio hasta n.

h(n) = costo estimado desde n hasta el destino (heurística).

La heurística más común en grids es la distancia Manhattan (abs(x1-x2) + abs(y1-y2)).

✅ Cuándo usarlo:

Cuando necesitas velocidad en mapas grandes.

Muy usado en juegos, GPS, IA de robots, etc.

#Para la reconstruccion del camino

Todos estos algorimto buscan la forma de llegar al final, pero para guardar el camino, hay que usar una estructura de NODO

# DIFERENCIA PRINCIPAL ENTRE DIJKSTRA Y A*

🔹 Dijkstra

Se centra en explorar todas las rutas posibles y elegir la más barata.

No “piensa” hacia dónde está el destino, simplemente avanza evaluando costos.

✅ Útil si:

Quieres garantizar siempre la ruta óptima.

El mapa es pequeño o mediano.

Los costos de los terrenos importan más que la velocidad de cálculo.

🔹 A*

Es básicamente Dijkstra + inteligencia: usa una heurística (ej: distancia Manhattan) para “dirigirse” hacia el destino.

Explora mucho menos que Dijkstra, porque prioriza los nodos que están más cerca del destino.

✅ Útil si:

El mapa es grande.

Quieres que la búsqueda sea rápida y eficiente.

Tu escenario es como un laberinto urbano donde hay que evitar perder tiempo explorando en todas direcciones.

# ¿Por que eelegi A*?

Entre Dijkstra y A*, ambos llegan al destino final, pero A* hace mas rapido, por que tiene en cuenta las coordenadas del final, y directamente se dirige hacia esas coordenadas

Implementar BFS termina siendo mucho mas lento, por que hace la busqueda hacia todas las direccinoes


# Pseudocódigo (esquemático):

    g[start] = 0
    push start en open con f = h(start)
    while open no vacío:
        n = pop nodo con menor f
        if n == goal: reconstruir ruta usando came_from
        marcar n como cerrado
        for cada vecino m de n:
            if m es obstáculo: continue
            tentative_g = g[n] + cost(n->m)
            if tentative_g < g[m] (mejor camino):
                came_from[m] = n
                g[m] = tentative_g
                f = tentative_g + h(m)
                push/update m en open


#Propiedades formales (completitud, optimalidad, complejidad)

Completitud: A* es completo en grafos finitos. En espacio infinito, es completo si existe un umbral positivo mínimo de coste por paso (no infinitesimales) y hay solución.

Optimalidad: si h es admisible, A* devuelve un camino óptimo. Si h además es consistente, la implementación estándar (no reabrir nodos) también es correcta.

Complejidad (informal):

Peor caso: puede ser exponencial en la profundidad del camino (p.ej. O(b^d)) — esto ocurre si la heurística es débil o nula.

Práctico: con una buena heurística la expansión suele ser muchísimo menor. En tableros finitos el coste dominante es: número de nodos expandidos × coste de operación del heap (log N), es decir O(N log N) en el peor caso donde N es número de nodos visitados.

# Detalles de implementación – errores comunes y buenas prácticas
a) Representación de costes

Decide con claridad: ¿el coste está asociado a la arista (moverse de A→B) o al nodo (entrar a B)?
En el ejemplo que ejecutamos, cost_to_enter = grid[nr][nc] (coste para entrar en la celda vecina). Es una convención simple y práctica.

b) Estructura de prioridad

Usa un heap (p.ej. heapq en Python).

Como heapq no permite “decrease-key”, la técnica común es: al encontrar mejor g para un nodo, meter otra entrada en el heap con su nuevo f. Al extraer, si la entrada extraída no coincide con el g_score actual (es un elemento stale), la descartas y sigues.

Añade un contador/tie-breaker al heap para evitar comparaciones entre tuplas iguales.

c) Manejo de igualdad de f (tie-breaking)

Cuando varios nodos tienen el mismo f, cómo rompes el empate afecta rendimiento:

Romper favorando mayor g (o equivalente: menor h) tiende a expandir nodos más “profundos” (más cerca del objetivo), y con frecuencia reduce el número de expansiones en grids.

Otra opción: tie-breaker por heurística (menor h) o por orden de inserción.

d) Consistencia vs reabrir nodos

Si la heurística es consistente, un nodo que ha sido movido al closed set nunca necesita reabrirse — esto simplifica el algoritmo. Si la heurística no es consistente, tendrás que permitir reabrir nodos (comprobar si un nuevo camino tiene g mejor y actualizar).

e) Comprobaciones prácticas

Asegúrate que todos los costes sean no negativos. Si hay coste 0 o muy pequeño, cuidado con ciclos (aunque con visited/cerrado se controla).

Si usas flotantes, compara con tolerancia (eps) para evitar bugs.

# Ejemplo paso a paso (conceptual, usando la ejecución que viste arriba)

En la ejecución que corrimos la heurística fue Manhattan y el coste de cada celda libre = 1.

start=(0,0), goal=(6,6) → h(start)=12, g(start)=0, f(start)=12.

Los vecinos inmediatos (si no son obstáculo) obtienen g=1, h=11, f=12. A* va sacando nodos con menor f (en ese mapa muchos tuvieron f=12), hasta que encuentra una ruta que rodea la muralla vertical y baja hasta la meta.

Resultado en la ejecución: ruta de coste 12 con 12 pasos (coincide con Manhattan porque el muro obliga a rodear pero no aumenta coste total por paso).

(Eso ilustra cómo A* combina g y h: cuando h lleva una buena estimación hacia el objetivo, muchas expansiones “irrelevantes” se evitan.)

# Errores frecuentes al implementar A*

Usar una heurística que sobreestima → puede perder optimalidad (entregará soluciones subóptimas).

No ajustar la heurística cuando los costes por paso cambian (p.ej. si el coste mínimo por paso > 1, escala la heurística por ese mínimo).

Manejar mal closed set cuando la heurística no es consistente → necesitas permitir reabrir nodos.

Olvidar filtrar entradas stale del heap (heapq no elimina automáticamente entradas antiguas).

Olvidar casos borde: start == goal, start/goal en obstáculo, coordenadas fuera del mapa.
