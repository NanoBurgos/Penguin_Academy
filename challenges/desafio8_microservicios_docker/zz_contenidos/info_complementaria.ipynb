## **Este desafio quiere que entiendas:**

**Microservicios**

Cada servicio es independiente.

**comunicación entre servicios**

            order → inventory

**autenticación**

Token Bearer.

**APIs REST**

Endpoints.

**contenedores**

Docker.

## **Investigar para poder explicar detalladamente:**

**arquitectura**: por qué hay 3 servicios.

**comunicación**: por qué order llama a inventory.

**docker**: por qué usamos docker compose.

**seguridad**: por qué usamos Bearer token.

**base de datos**: por qué cada servicio tiene su propia DB.

## arrancar docker 

abrir docker desktop, luego en visual, en terminal 

            docker-compose up --build

Paso 1: Crear producto
POST http://localhost:8001/products

Body:

{
  "name": "Laptop",
  "stock": 10
}

✔ Resultado esperado:

Producto creado con ID

🔹 Paso 2: Ver producto
GET http://localhost:8001/products

✔ Debe aparecer tu producto

🔹 Paso 3: Crear orden
POST http://localhost:8002/orders
{
  "product_id": 1,
  "quantity": 2
}
🚨 5. Validación crítica (LO MÁS IMPORTANTE)

Después de crear la orden:

👉 vuelve a consultar inventario

GET http://localhost:8001/products
✔ Correcto:
stock: 8
❌ Incorrecto:
stock: 10

👉 Si pasa esto:
tu sistema está mal diseñado (no descuenta stock)

🔍 6. Ver logs (clave en microservicios)

Ejecuta:

docker-compose logs -f

O por servicio:

docker logs inventory_service
docker logs order_service

👉 Aquí debes ver:

Requests entrando

Validaciones

Llamadas entre servicios

🔁 7. Probar comunicación interna (MUY IMPORTANTE)

Tu order_service debe llamar a:

http://inventory_service:8000

NO:

http://localhost:8001

👉 Esto solo funciona fuera de Docker

⚠️ 8. Simular errores reales

Prueba estos casos:

❌ Comprar más stock del disponible
{
  "product_id": 1,
  "quantity": 999
}

✔ Debe responder:

error: insufficient stock
❌ Sin token

✔ Debe dar:

401 Unauthorized
❌ Producto inexistente

✔ Debe dar:

404 Not Found
