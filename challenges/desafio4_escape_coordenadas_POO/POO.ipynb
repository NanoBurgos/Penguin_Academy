#**Los 4 pilares de la Programación Orientada a Objetos (POO)**

#Abstracción 🧩

    Consiste en modelar los objetos del mundo real o conceptos en clases, mostrando solo lo esencial y ocultando los detalles innecesarios.

    Ejemplo: una clase Auto abstrae características como color, marca
    y métodos como arrancar(), sin mostrar al usuario cómo funciona
    el motor internamente.

#Encapsulamiento 🔒

    Significa ocultar los datos internos de un objeto y controlar su acceso a través de métodos (getters, setters o propiedades).

    Protege la integridad de los datos y permite modularidad.

    Ejemplo: los atributos de CuentaBancaria son privados y solo se accede mediante métodos que validan operaciones.

#Herencia 🧬

    Permite que una clase (hija) herede atributos y métodos de otra clase (padre).

    Facilita la reutilización de código y el uso de jerarquías.

    Ejemplo: una clase Perro puede heredar de una clase Animal propiedades como edad y métodos como comer().

#Polimorfismo 🎭

    Significa que un mismo método puede tener diferentes comportamientos según el objeto que lo invoque.

    Permite que métodos con el mismo nombre funcionen de manera distinta en distintas clases.

    Ejemplo: Animal.sonido() puede hacer que un perro ladre y un gato maúlle, aunque ambos usen el mismo método.

Separamos en Propiedades(Atributos) y Funciones(Metodos)

**Atributos** = son las propiedades que definen al objeto

**Metodos** = las acciones que pueden realizar

# Ej de una Taza

**Atributos**: volumen(total), liquido(con el que se rellena), cantidad(de ese liquido)

**Metodos**: llenar(), beber(), vaciar()

y para especificar que queremos crear un tipo de dato nuevo, se usa la clase

La clase es la plantilla donde definimos los atributos y metodos


Para instanciar una clase, se usa el metodo constructor

**El metodo constructor** es un metodo que sirve para crear el objeto y darle los valores necesarios para iniciarlo

El nombre del metodo constructor depende del lenguaje utilizado

En python: __init__
En Java: el mismo nombre de la clase

Una vez creado el objeto, podemos acceder con un punto(.) a sus atributos ya sea para consultar o modificar o para utilizar sus metodos

Con la **Abstraccion**, elegimos cuales seran los atributos y los metodos que definiran a nuestra clase


Con la **Encapsulacion** es poner algo en privado las cosas que no queremos que se ejecuten desde afuera, Desde afuera no serian accesibles con el punto(.)

get -> acceder

set -> modificar

En resumen,si no declaramos los atributos como privados, podemos usar esos atributos desde fuera sin ningun tipo de control. Pero al declararlo como privado, estos no podran consultarse ni modificarse, a no ser que usemos los metodos get y set



Con la **Herencia**, por ej si en un juego, tenemos la clase personaje, y ahora queremos crear personajes, pero con algunas diferencias y/o atributos nuevos y metodos nuevos; En vez de crear todo desde cero, podemos reutilizar la clase personaje con la **herencia**

A la clase que hereda se le conoce como, *clase Padre o Super Clase* y a la clase heredade se le conoce como *clase Hija o Subclase*



Con el **Polimorfismo**, esto permite a un metodo ser diferente, atendiendo que clase lo este usando, Es decir puede tener muchas formas de usarse

```python
"""Ahora vamos a crear un metodo que devuelva un valor boolean que indique si
el personaje esta vivo o no"""

class Personaje:

  def __init__(self, nombre, fuerza, inteligencia, defensa, vida):
    self.nombre = nombre
    self.fuerza = fuerza
    self.inteligencia = inteligencia
    self.defensa = defensa
    self.vida = vida

  def atributos(self):
    print(self.nombre, ":", sep="")
    print("·Fuerza:", self.fuerza)
    print("·Inteligencia:", self.inteligencia)
    print("·Defensa:", self.defensa)
    print("·Vida:", self.vida)

  def subir_nivel(self, fuerza, inteligencia, defensa, vida):
    self.fuerza = self.fuerza + fuerza
    self.inteligencia = self.inteligencia + inteligencia
    self.defensa = self.defensa + defensa
    self.vida = self.vida + vida

  def esta_vivo(self):
    return self.vida > 0

print("Primero creamos al personaje con estos atributos (Isaac Netero, 60, 75, 80, 25)")
mi_personaje = Personaje("Isaac Netero", 60, 75, 80, 25)
mi_personaje.atributos()
print(mi_personaje.esta_vivo())

print("\nLuego vamos a hacerle subir de nivel (Isaac Netero, +5, +10, +8, -25)")
mi_personaje.subir_nivel(5, 10, 8, -25)
mi_personaje.atributos()
print(mi_personaje.esta_vivo())
```

```python
"""Ahora vamos a crear un metodo que represente la accion de
morir del personaje, es decir, ponga su vida en cero, y nos
indique que ha muerto """

class Personaje:

  def __init__(self, nombre, fuerza, inteligencia, defensa, vida):
    self.nombre = nombre
    self.fuerza = fuerza
    self.inteligencia = inteligencia
    self.defensa = defensa
    self.vida = vida

  def atributos(self):
    print(self.nombre, ":", sep="")
    print("·Fuerza:", self.fuerza)
    print("·Inteligencia:", self.inteligencia)
    print("·Defensa:", self.defensa)
    print("·Vida:", self.vida)

  def subir_nivel(self, fuerza, inteligencia, defensa, vida):
    self.fuerza = self.fuerza + fuerza
    self.inteligencia = self.inteligencia + inteligencia
    self.defensa = self.defensa + defensa
    self.vida = self.vida + vida

  def esta_vivo(self):
    return self.vida > 0

  def morir(self):
    self.vida = 0
    print(self.nombre, "Ha muerto")

print("Primero creamos al personaje con estos atributos (Isaac Netero, 60, 75, 80, 25)")
mi_personaje = Personaje("Isaac Netero", 60, 75, 80, 25)
mi_personaje.atributos()
print(mi_personaje.esta_vivo())

print("\nLuego vamos a probar el metodo morir")
print("")
mi_personaje.morir()

print("")

#Y comprobamos los atributos para verificar que la vida a cambiado a cero
mi_personaje.atributos()

```

```python
"""Ahora vamos a crear un metodo que calcule el daño que realiza nuestro personaje a otro personaje
Para esto hay que crear otro personaje que represente a nuestro enemigo, pero podemos hacerlo usando la clase Personaje

 Tambien vamos a crear el metodo atacar, que resta el daño a la vida del enemigo
 """
class Personaje:

  def __init__(self, nombre, fuerza, inteligencia, defensa, vida):
    self.nombre = nombre
    self.fuerza = fuerza
    self.inteligencia = inteligencia
    self.defensa = defensa
    self.vida = vida

  def atributos(self):
    print(self.nombre, ":", sep="")
    print("·Fuerza:", self.fuerza)
    print("·Inteligencia:", self.inteligencia)
    print("·Defensa:", self.defensa)
    print("·Vida:", self.vida)

  def subir_nivel(self, fuerza, inteligencia, defensa, vida):
    self.fuerza = self.fuerza + fuerza
    self.inteligencia = self.inteligencia + inteligencia
    self.defensa = self.defensa + defensa
    self.vida = self.vida + vida

  def esta_vivo(self):
    return self.vida > 0

  def morir(self):
    self.vida = 0
    print(self.nombre, "Ha muerto")

  def daño(self, enemigo):
    return self.fuerza - enemigo.defensa

  def atacar(self, enemigo):
    daño = self.daño(enemigo)
    enemigo.vida = enemigo.vida - daño
    print(self.nombre, "ha realizado", daño, "puntos de daño a", enemigo.nombre)
    if enemigo.esta_vivo():
      print("La vida de", enemigo.nombre, "es", enemigo.vida)
    else:
      enemigo.morir()

print("Primero creamos al personaje con estos atributos (Isaac Netero, 60, 75, 80, 25)")
mi_personaje = Personaje("Isaac Netero", 85, 75, 80, 25)
print("Despues creamos al enemigo con estos atributos (Meruem, 80, 100, 80, 95))")
mi_enemigo = Personaje("Meruem", 80, 100, 80, 95)

#Probamos para ver como funciona el metodo
print(mi_personaje.daño(mi_enemigo))

print("")
#Probamos para ver como funciona el metodo atacar, e imprimimos los atributos para ver si actualiza
mi_personaje.atacar(mi_enemigo)
print("")
mi_enemigo.atributos()
```
